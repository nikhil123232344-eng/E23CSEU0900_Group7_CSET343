```python
print("Missing serum_creatinine values:",
      df["serum_creatinine"].isnull().sum())
```

```python
df["serum_creatinine"] = df["serum_creatinine"].fillna(
    df["serum_creatinine"].median()
)
```

```python
print("Missing smoking values:",
      df["smoking"].isnull().sum())
```

```python
df["smoking"] = df["smoking"].fillna(
    df["smoking"].mode()[0]
)
```

```python
print("Total missing values after imputation:",
      df.isnull().sum().sum())
```

```python
def remove_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    return data[
        (data[column] >= lower_bound) &
        (data[column] <= upper_bound)
    ]
```

```python
print("Rows before removing outliers:", df.shape[0])
```

```python
df = remove_outliers_iqr(df, "age")
df = remove_outliers_iqr(df, "ejection_fraction")
```

```python
print("Rows after removing outliers:", df.shape[0])
```

```python
print("Minimum age:", df["age"].min())
print("Maximum age:", df["age"].max())
```

```python
print("Negative ages:", (df["age"] < 0).sum())
print("Negative ejection fractions:", (df["ejection_fraction"] < 0).sum())
print("Negative serum creatinine:", (df["serum_creatinine"] < 0).sum())
```

```python
df.loc[df["age"] < 0, "age"] = df["age"].median()
```

```python
print("Negative ages after correction:",
      (df["age"] < 0).sum())
```

```python
def age_group(age):
    if age < 40:
        return "Young"
    elif age <= 60:
        return "Middle"
    else:
        return "Old"

df["age_group"] = df["age"].apply(age_group)
```

```python
df[["age", "age_group"]].head(10)
```

```python
df["risk_score"] = (
    df["ejection_fraction"] * df["serum_creatinine"]
)
```

```python
min_value = df["risk_score"].min()
max_value = df["risk_score"].max()

df["risk_score"] = (
    (df["risk_score"] - min_value) /
    (max_value - min_value)
)
```
