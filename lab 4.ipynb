```python
# Step 1: Import required libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    roc_auc_score
)

print("All libraries imported successfully!")
```

```python
# Step 2: Load the Breast Cancer Wisconsin dataset

from sklearn.datasets import load_breast_cancer

# Load dataset
data = load_breast_cancer()

# Create DataFrame
df = pd.DataFrame(data.data, columns=data.feature_names)

# Add target column
df["Outcome"] = data.target

# Display first 5 rows
df.head()
```

```python
# Check the shape of the dataset

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
# Display column names

print("Features:")
for column in df.columns:
    print(column)
```

```python
# Step 3: Summary statistics

df.describe().T
```

```python
# Step 4: Check for missing values

print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
# Check for duplicate rows

print("Number of duplicate rows:", df.duplicated().sum())
```

```python
# Remove duplicate rows if any

df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
# Step 5: Detect outliers using IQR method

# Select only feature columns (exclude Outcome)
features = df.drop(columns=["Outcome"])

Q1 = features.quantile(0.25)
Q3 = features.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Count outliers in each feature
outlier_counts = ((features < lower_bound) | (features > upper_bound)).sum()

print("Outliers in each feature:")
print(outlier_counts)
```

```python
# Total number of outlier values

print("Total outlier values:", outlier_counts.sum())
```

```python
# Box plots for selected important features

plt.figure(figsize=(12, 6))

df[[
    "mean radius",
    "mean texture",
    "mean perimeter",
    "mean area",
    "mean smoothness"
]].boxplot()

plt.title("Box Plot of Selected Features")
plt.ylabel("Values")
plt.xticks(rotation=45)
plt.show()
```

```python
# Step 6.1: Histograms of selected features

selected_features = [
    "mean radius",
    "mean texture",
    "mean perimeter",
    "mean area",
    "mean smoothness"
]

df[selected_features].hist(figsize=(12, 8), bins=20)

plt.suptitle("Distribution of Selected Features")
plt.tight_layout()
plt.show()
```

```python
# Step 6.2: Correlation heatmap

plt.figure(figsize=(16, 12))

correlation_matrix = df.corr()

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    annot=False
)

plt.title("Correlation Heatmap of Features")
plt.show()
```

```python
# Step 6.3: Class distribution

print("Class distribution:")
print(df["Outcome"].value_counts())

plt.figure(figsize=(6, 5))

sns.countplot(x="Outcome", data=df)

plt.title("Class Distribution")
plt.xlabel("Outcome (0 = Malignant, 1 = Benign)")
plt.ylabel("Number of Samples")

plt.show()
```

```python
# Class percentages

class_percentage = df["Outcome"].value_counts(normalize=True) * 100

print("Class percentages:")
print(class_percentage)
```

```python
# Step 7.1: Analyze class distribution

class_counts = df["Outcome"].value_counts()

print("Class Counts:")
print(class_counts)

print("\nClass Percentages:")
print((class_counts / len(df) * 100).round(2))
```

```python
# Step 7.2: Separate features (X) and target (y)

X = df.drop(columns=["Outcome"])
y = df["Outcome"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# Step 7.3: Train-test split using stratified sampling

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
# Check class distribution in training and testing sets

print("Training class distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting class distribution:")
print(y_test.value_counts(normalize=True).round(3))
```

```python
# Step 8: Standardize the features

scaler = StandardScaler()

# Fit only on training data and transform training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform test data using the same scaler
X_test_scaled = scaler.transform(X_test)

print("Training data shape:", X_train_scaled.shape)
print("Testing data shape:", X_test_scaled.shape)
```

```python
# Check mean and standard deviation after scaling

print("Mean of scaled training features:")
print(X_train_scaled.mean(axis=0)[:5])

print("\nStandard deviation of scaled training features:")
print(X_train_scaled.std(axis=0)[:5])
```

```python
scaler.fit_transform(X_train)
```

```python
scaler.transform(X_test)
```

```python
# Step 9: Train Logistic Regression model

model = LogisticRegression(random_state=42, max_iter=1000)

# Train the model
model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully!")
```

```python
# Make predictions on test data

y_pred = model.predict(X_test_scaled)

print("Predictions:")
print(y_pred[:20])
```

```python
# Get probability predictions

y_pred_proba = model.predict_proba(X_test_scaled)[:, 1]

print("Prediction probabilities:")
print(y_pred_proba[:10])
```

```python
# Step 10: Evaluate Logistic Regression model

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average=None)
recall = recall_score(y_test, y_pred, average=None)
f1 = f1_score(y_test, y_pred, average=None)

print("Model Evaluation Results")
print("-" * 30)

print(f"Accuracy: {accuracy:.4f}")

print("\nPrecision:")
print(f"Class 0 (Malignant): {precision[0]:.4f}")
print(f"Class 1 (Benign):    {precision[1]:.4f}")

print("\nRecall:")
print(f"Class 0 (Malignant): {recall[0]:.4f}")
print(f"Class 1 (Benign):    {recall[1]:.4f}")

print("\nF1-Score:")
print(f"Class 0 (Malignant): {f1[0]:.4f}")
print(f"Class 1 (Benign):    {f1[1]:.4f}")
```

```python
# Classification report

print("Classification Report")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Malignant", "Benign"]
))
```

```python
# Step 11: Confusion Matrix

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
# Visualize Confusion Matrix

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Malignant", "Benign"],
    yticklabels=["Malignant", "Benign"]
)

plt.title("Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.show()
```

```python
# Step 12: ROC Curve and AUC Score

# Calculate AUC
auc_score = roc_auc_score(y_test, y_pred_proba)

# Calculate ROC curve
fpr, tpr, thresholds = roc_curve(y_test, y_pred_proba)

print("ROC-AUC Score:", round(auc_score, 4))
```

```python
# Plot ROC Curve

plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    label=f"Logistic Regression (AUC = {auc_score:.4f})"
)

# Random classifier line
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.grid()

plt.show()
```
