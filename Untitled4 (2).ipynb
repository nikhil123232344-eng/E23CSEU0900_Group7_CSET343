```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
```

```python
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"

columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
    "Outcome"
]

df = pd.read_csv(url, names=columns)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()
```

```python
# Check actual missing values
print("Missing values (NaN):")
print(df.isnull().sum())

# Check zero values in each column
print("\nZero values:")
print((df == 0).sum())
```

```python
# Replace invalid zero values with NaN
invalid_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

df[invalid_columns] = df[invalid_columns].replace(0, np.nan)

# Check missing values again
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# Fill missing values with the median
for col in invalid_columns:
    df[col] = df[col].fillna(df[col].median())

# Verify that no missing values remain
print("Missing values after median imputation:")
print(df.isnull().sum())
```

```python
# Descriptive statistics
df.describe()
```

```python
print("Outcome Counts:")
print(df["Outcome"].value_counts())

# Visualize the outcome distribution
sns.countplot(x="Outcome", data=df)

plt.title("Diabetes Outcome Distribution")
plt.xlabel("Outcome (0 = Non-Diabetic, 1 = Diabetic)")
plt.ylabel("Number of Patients")
plt.show()
```

```python
# Compare important clinical features by Outcome

features = ["Glucose", "BloodPressure", "BMI", "Age"]

for feature in features:
    plt.figure(figsize=(7, 4))
    sns.boxplot(x="Outcome", y=feature, data=df)
    plt.title(f"{feature} by Diabetes Outcome")
    plt.xlabel("Outcome (0 = Non-Diabetic, 1 = Diabetic)")
    plt.ylabel(feature)
    plt.show()
```

```python
# Calculate correlation matrix
correlation = df.corr()

print(correlation)

# Visualize correlations
plt.figure(figsize=(10, 7))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Matrix of Pima Diabetes Dataset")
plt.show()
```

```python
# Separate glucose values by diabetes outcome
glucose_non_diabetic = df[df["Outcome"] == 0]["Glucose"]
glucose_diabetic = df[df["Outcome"] == 1]["Glucose"]

# Independent two-sample t-test
t_stat, p_value = stats.ttest_ind(
    glucose_non_diabetic,
    glucose_diabetic,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)

# Decision
if p_value < 0.05:
    print("Reject H0: There is a significant difference in mean glucose levels.")
else:
    print("Fail to reject H0: No significant difference in mean glucose levels.")
```

```python
# Create a categorical pregnancy variable
df["Pregnancy_Status"] = np.where(
    df["Pregnancies"] == 0,
    "No Pregnancy",
    "At Least One Pregnancy"
)

# Create contingency table
contingency_table = pd.crosstab(
    df["Pregnancy_Status"],
    df["Outcome"]
)

print("Contingency Table:")
print(contingency_table)

# Chi-square test
chi2, p_value, dof, expected = stats.chi2_contingency(contingency_table)

print("\nChi-square statistic:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", dof)

# Decision
if p_value < 0.05:
    print("Reject H0: Pregnancy status and diabetes outcome are significantly associated.")
else:
    print("Fail to reject H0: No significant association was found.")
```

```python
# Create pregnancy groups
df["Pregnancy_Group"] = pd.cut(
    df["Pregnancies"],
    bins=[-1, 0, 3, 6, np.inf],
    labels=["0", "1-3", "4-6", "7+"]
)

# Create groups of glucose values
groups = [
    group["Glucose"].values
    for _, group in df.groupby("Pregnancy_Group", observed=True)
]

# Perform one-way ANOVA
f_stat, p_value = stats.f_oneway(*groups)

print("F-statistic:", f_stat)
print("P-value:", p_value)

# Decision
if p_value < 0.05:
    print("Reject H0: Mean glucose levels differ significantly among pregnancy groups.")
else:
    print("Fail to reject H0: No significant difference among pregnancy groups.")
```

```python
# Feature 1: Age group
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0, 30, 50, np.inf],
    labels=["Young", "Middle-Aged", "Older"]
)

# Feature 2: BMI category
df["BMI_Category"] = pd.cut(
    df["BMI"],
    bins=[0, 18.5, 25, 30, np.inf],
    labels=["Underweight", "Normal", "Overweight", "Obese"]
)

# Feature 3: Glucose category
df["Glucose_Category"] = pd.cut(
    df["Glucose"],
    bins=[0, 100, 125, np.inf],
    labels=["Normal", "Prediabetes", "High"]
)

# Display the new features
df[["Age", "Age_Group", "BMI", "BMI_Category",
    "Glucose", "Glucose_Category"]].head()
```

```python
# Correlation of numerical features with Outcome
correlation_with_outcome = df.corr(numeric_only=True)["Outcome"].sort_values(
    ascending=False
)

print("Correlation with Diabetes Outcome:")
print(correlation_with_outcome)
```

```python
from sklearn.preprocessing import StandardScaler

# Select numerical features
numerical_features = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age"
]

# Standardize the features
scaler = StandardScaler()

df_scaled = df.copy()

df_scaled[numerical_features] = scaler.fit_transform(
    df_scaled[numerical_features]
)

# Display the scaled data
df_scaled.head()
```

```python
# Check mean and standard deviation after scaling

print("Means after scaling:")
print(df_scaled[numerical_features].mean())

print("\nStandard deviations after scaling:")
print(df_scaled[numerical_features].std())
```

```python
# Detect outliers using IQR

for col in numerical_features:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) | (df[col] > upper_bound)
    ]

    print(f"{col}: {len(outliers)} outliers")
```
